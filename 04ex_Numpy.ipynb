2\. Find the outer product of the following two vecotrs

```python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
```

Do this in the following ways:

   * Using the function outer in numpy
   * Using a nested for loop or list comprehension
   * Using numpy broadcasting operatoins


````python
# Using the function outer in numpy:
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
M1 = np.outer(u,v)

# Using a nested for loop or list comprehension:
M2 = np.array([[ui * vj for vj in v] for ui in u]) 

# Using numpy broadcasting operations:
M3 = u[:,np.newaxis] * v # “ np.newaxis " adds a new dimension of size 1 to a NumPy array, changing its shape.

print(M1,'\n','\n',M2,'\n','\n',M3)
````

4\. Use np.linspace to create an array of 100 numbers between 0 and 2π (includsive).

  * Extract every 10th element using slice notation
  * Reverse the array using slice notation
  * Extract elements where the absolute difference between the sine and cosine functions evaluated at that element is less than 0.1
  * Make a plot showing the sin and cos functions and indicate where they are close

````python
import matplotlib.pyplot as plt
E = np.linspace(0,2 * np.pi,100)

# We extract every 10th element.
mask = np.arange(9,len(E),10)
tenth = E[mask]
print("Every 10th number:",tenth,'\n')

# We reverse the array.
E_reversed = E[::-1]
print("Array reversed:",E_reversed,'\n')

# We extract elements where the absolute difference between the sine and cosine functions evaluated at that element is less than 0.1 .
dif = abs(np.sin(E)-np.cos(E))
E_low = E[0.1 > dif] 
print("Lower difference values:",E_low,'\n')

# We make a plot showing the sin and cos functions indicating where are they close.
print("Plot showing the sin and cos functions indicating where are they close",'\n')
plt.plot(E,np.sin(E),label = 'sin(x)')
plt.plot(E,np.cos(E),label = 'cos(x)')
for point in E_low:
    plt.axvline(x=point, color='r', linestyle='-')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.show()
````

5\. Create a matrix that shows the 10 by 10 multiplication table.

 * Find the trace of the matrix
 * Extract the anti-diagonal (this should be ```array([10, 18, 24, 28, 30, 30, 28, 24, 18, 10])```)
 * Extract the diagonal offset by 1 upwards (this should be ```array([ 2,  6, 12, 20, 30, 42, 56, 72, 90])```)

````python
Mult = np.array([[i*j for j in range(1,11)] for i in range(1,11)])
print(Mult,'\n')

# Trace.
Diag = np.diag(Mult)
Trace = sum(Diag)
print("Trace:",Trace,'\n')

# Anto-diagonal.
Anti_diag = [Mult[9-i,i] for i in range(10)]
print("Anti-diagonal:",Anti_diag,'\n')

# Diagonal offset by 1 upwards.
Diag_up = [Mult[i,i+1] for i in range(9)]
print("Diagnoal offset by 1 upwards:",Diag_up,'\n')
````

6\. Use broadcasting to create a grid of distances

Route 66 crosses the following cities in the US: Chicago, Springfield, Saint-Louis, Tulsa, Oklahoma City, Amarillo, Santa Fe, Albuquerque, Flagstaff, Los Angeles
The corresponding positions in miles are: 0, 198, 303, 736, 871, 1175, 1475, 1544, 1913, 2448

  * Construct a 2D grid of distances among each city along Route 66
  * Convert that in km (those savages...)

````python
# 2D grid of distances among each city along Route 66.
Miles = np.array([0, 198, 303, 736, 871, 1175, 1475, 1544, 1913, 2448])
Distances = np.array([[abs(Miles[i] - Miles[j]) for j in range(10)]for i in range(10)])
# Or: Distances = np.abs(Miles[:, None] - Miles[None, :])
print('2D grid:','\n',Distances,'\n')

# Distances in km.
print('2D grid in km:','\n',np.round(Distances * 1.61),'\n')
````

7\. Prime numbers sieve: compute the prime numbers in the 0-N (N=99 to start with) range with a sieve (mask).
  * Constract a shape (100,) boolean array, the mask
  * Identify the multiples of each number starting from 2 and set accordingly the corresponding mask element
  * Apply the mask to obtain an array of ordered prime numbers
  * Check the performances (timeit); how does it scale with N?
  * Implement the optimization suggested in the [sieve of Eratosthenes](https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes)

````python
# Prime numbers in the 0-N range with a sieve (mask).

# Shape (100,) boolean array, the mask.
def primes(N):
    is_prime = np.ones(N, dtype=bool) # First we asume that all the numbers are prime except 0 and 1.
    is_prime[0] = False
    is_prime[1] = False
    
# Multiples of each number starting from 2 and set accordingly the corresponding mask element.
    for i in range(2, N):
        is_prime[i*i:N:i] = False
    return np.arange(N)[is_prime]
    
# Ordered prime numbers.
print("Primes:",'\n',primes(100)) # 100 is the input because the mask gives 100 values starting on 0.

# Performances (timeit).
%timeit primes(100)
%timeit primes(200)
%timeit primes(300)

# Optimization suggested in the sieve of Eratosthenes.
def primes_op(N):
    is_prime = np.ones(N, dtype=bool) 
    is_prime[0] = False
    is_prime[1] = False
    for i in range(2, int(np.sqrt(N)) + 1):
        if is_prime[i]: # This makes it faster because doesn't repeat non-prime numbers.
            is_prime[i*i:N:i] = False
    return np.arange(N)[is_prime]

print("Primes optimized:",'\n',primes_op(100))
%timeit primes_op(100)
%timeit primes_op(500)
%timeit primes_op(1000)
````

8\. Diffusion using random walk

Consider a simple random walk process: at each step in time, a walker jumps right or left (+1 or -1) with equal probability. The goal is to find the typical distance from the origin of a random walker after a given amount of time. 
To do that, let's simulate many walkers and create a 2D array with each walker as a raw and the actual time evolution as columns

  * Take 1000 walkers and let them walk for 200 steps
  * Use randint to create a 2D array of size walkers x steps with values -1 or 1
  * Build the actual walking distances for each walker (i.e. another 2D array "summing on each raw")
  * Take the square of that 2D array (elementwise)
  * Compute the mean of the squared distances at each step (i.e. the mean along the columns)
  * Plot the average distances (sqrt(distance\*\*2)) as a function of time (step)
  
Did you get what you expected?

````python
import matplotlib.pyplot as plt
# Take 1000 walkers and let them walk for 200 steps.
N = 1000
t = 200

# Use randint to create a 2D array of size walkers x steps with values -1 or 1.
M = np.random.randint(0,2,N*t).reshape(N,t) * 2 - 1

# Walking distances.
Distances = np.cumsum(M,axis=1) # This is the cumulative distances.

# Square of the array.
D2 = Distances**2

# Mean of the squared distances at each step.
Mean_dist = np.mean(D2, axis=0)

# Average distances as a function of time (step).
plt.plot(np.arange(1,t+1),Mean_dist**0.5, label='Average distances')
plt.plot(np.arange(1,t+1), np.sqrt(np.arange(1,t+1)), '--', label=r'$\sqrt{t}$')
plt.legend()
plt.title("Random Walker")
plt.xlabel("Steps")
plt.ylabel("Distance")
print("As we expected, the numerical results show that the root mean square displacement increases proportionally to the square root of the number of steps, as expected for a simple random walk.")
````

  * Check the content by looking within the file

  * Create arrays out of the columns, the arrays being (in order): *year*, *hares*, *lynxes*, *carrots* 
  * Plot the 3 populations over the years

  * Which species has the highest population each year?

Do you feel there is some evident correlation here? [Studies](https://www.enr.gov.nt.ca/en/services/lynx/lynx-snowshoe-hare-cycle) tend to believe so.

````python
# Load the data.
data = np.loadtxt("populations.txt")

# Create arrays out of the columns, the arrays being (in order): year, hares, lynxes, carrots.
years = data[:,0]
hares = data[:,1]
lynxes = data[:,2]
carrots = data[:,3]

# The 3 populations over the years plotted.
plt.plot(years,hares,label="hares")
plt.plot(years,lynxes,label="lynxes")
plt.plot(years,carrots,label="carrots")
plt.legend()
plt.xlabel("Years")
plt.ylabel("Population")

# Statistics.
print("Means of hares, lynxes and carrots, respectively:", np.mean(data[:,1:], axis=0),'\n')
print("Std of hares, lynxes and carrots, respectively:", np.std(data[:,1:], axis=0),'\n')
print("Correlations:\n", np.corrcoef(data[:,1:].T),'\n')

# Dominant species each year
species = np.array(["hares", "lynxes", "carrots"])
max_species = species[np.argmax(data[:,1:], axis=1)] # " np.argmax " gives the position (max would give the value).
Dominant_population = np.column_stack((years, max_species))
print('Highest population each year:','\n',Dominant_population)
````
